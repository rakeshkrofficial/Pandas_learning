# Series From List

```python
# String

country = ["India", "Russia", "Nepal", "Japan", "Srilanka"]

pd.Series(country)
```

```python
# Integer

Mahi_runs = [57,98,78,49,100]

pd.Series(Mahi_runs)
```

```python
# Custom index

marks = [89,93,87,91,82,72]
subjects = ["Social science","Mathematics","Science","Hindi","Sanskrit","English"]

pd.Series(marks,index=subjects)
```

```python
pd.Series(marks,index=subjects,name="My 10th marks sheet")
```

# Editing Series

```python
marks_series = [89,93,87,91,82,72]
subjects = ["Social science","Mathematics","Science","Hindi","Sanskrit","English"]

marks=pd.Series(marks_series,index=subjects)
marks
```

```python
# using indexing
marks['Mathematics']=100
marks
```

```python
# what if an index doest not exist

marks['computer science']=85
marks
```

```python
# slicing

marks[1:4]=[98,90,97]
marks
```

```python
# fancy indexing
boly = boly[~boly.index.duplicated()] # in boly has dublicate index so use it
boly[['Zokkomon','Zindagi 50-50','Zor Lagaa Ke...Haiya!']]=['Darsheel Safary sir','Veena Malik sir','Meghan Jadhav sir']
boly
```

```python
number[[0,3,5]]=[0,0,0]
number
```

```python
# using index label

boly['Zokkomon'] ='Darsheel Safary'
boly
```
